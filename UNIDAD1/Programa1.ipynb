#**Listas en python**

```python
#Podemos acceder a cada elemento o item de la lista a traves
#del indice o su posicion. Las listas siempre empiezan en la posicion 0
alumnos=["Ana","Luis","Carlos"]
print(alumnos)
print(alumnos[0])
print(alumnos[2])
```

```python
#Podemos modificar el valor de un elemento de la lista
#Usando el nombre de la lista y el indice y le asignamos el nuevo
print(alumnos)
alumnos[1]="Juan"
print(alumnos)
```

```python
#Podemos especificar indices negativos y cuando se ponen valores
#negativos se lee o se accede a la lista desde el final
#empezando con -1
print(alumnos)
print(alumnos[-1])
print(alumnos[-2])
print(alumnos[-3])
```

```python
# La funcion append nos sirve para agregar un elemento al final de la lista
print(alumnos)
alumnos.append("Sandra")
print(alumnos)
```

```python
print(alumnos)
alumnos.insert(2,"Sofia")
print(alumnos)
```

```python
#Revome borra un elemento de la lista especificando el valor
#si excisten valores repetidos, borra el primer valor que encuentra.
print(alumnos)
alumnos.remove("Sofia")
print(alumnos)
```

```python
print(alumnos[2])
#La siguiente linea marca error  por que queremos acceder a
#un elemento que no existe.
print(alumnos[7])
```

```python
#Pop me permite eliminar un elemento especifico de la lista
#especificando su posicion
print(alumnos)
alumnos.pop(2)
print(alumnos)
```

```python
#Del me permite eliminar un elemento especifico de la lista
#especificando su posicion
print(alumnos)
del alumnos[3]
print(alumnos)
```

```python
#La funcion len me pemite conocer el numero de elementos en la lista.
print("Número de elementos o items en la lista", len(alumnos))
```

```python
#Ciclo para recorrer una lista y hacer un proceso
alumnos.append("Sandra")
alumnos.append("Carlos")
print(alumnos)
for x in alumnos:
  if x=="Sofia":
    print("Encontramos a Sofia")
    break #Break forza a salir de un ciclo
  print(x)
```
