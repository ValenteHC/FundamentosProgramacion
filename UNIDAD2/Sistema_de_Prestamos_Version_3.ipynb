```python
#========================================
# SISTEMA DE PRÉSTAMOS
# Versión 3 - Programación basica y uso de listas
#========================================

#Lista de alumnos activos en el sistema
alumnos = ["26040200","26040201","26040202","26040203","26040204"]

# Cada valor en la lista de computadoras tiene su lista correspondiente con el
# aula donde esta ubicada, el software instalado y el estado de esa computadora
computadoras = [
               [ "PC-LC1-01","LC1","Python, VS Code","Disponible"],
               [ "PC-LC1-02","LC1","Python, Java, VS Code","Prestada" ],
               [ "PC-LC1-03","LC1","AutoCAD, MATLAB","Reservada" ],
               [ "PC-LC2-01","LC2","Python, VS Code", "Disponible" ],
               [ "PC-LC2-02","LC2","Python, Java, VS Code","Prestada" ],
               [ "PC-LC2-03","LC2", "AutoCAD, MATLAB", "Reservada" ],
               [ "PC-LC3-01","LC3","Python, VS Code", "Disponible" ],
               [ "PC-LC3-02","LC3", "Python, Java, VS Code", "Prestada" ],
               [ "PC-LC3-03","LC3", "AutoCAD, MATLAB", "Reservada" ],
               [ "PC-LC3-04","LC3",  "Python, VS Code", "Mantenimiento" ]
            ]

#1
def listar_computadoras():
  print("\n---COMPUTADORAS---")
  for i in range(len(computadoras)):
      print(i+1 , " | Aula:", computadoras[i][1],"| ID:", computadoras[i][0],"| Software:", computadoras[i][2],"| Estado:", computadoras[i][3],"|")
      print("---------------------------------------------------------------------------------")


#Las triples comillas nos sirven para hacer un comentario las partes del codigo que deseamos,
#necesitamos poner las comillas al inicio y al final de lo que deseamos poner como comentario

#2
def buscar_sw():
  programa = input( "\nIngrese el software que desea buscar: ")
  print("\n--- RESULTADOS ---")
  encontrado = False
  for i in range(len(computadoras)):
      # la funcion (lower) convierte a minuscula el texto, de esta forma
      # nos aseguramos que el texto a buscar y la informacion de las listas
      # este en minuscula y no tengamos problema de que no encunentra un programa
      # porque este escrito de forma diferente usando mayusculas y minusculas
      if programa.lower() in computadoras[i][2].lower():
          print(i, " | Aula:", computadoras[i][1],"| ID:", computadoras[i][0],"| Software:", computadoras[i][2],"| Estado:", computadoras[i][3],"|")
          print("---------------------------------------------------------------------------------")
          encontrado = True
  if encontrado == False:
      print("No se encontraron computadoras con ese software.")

#3
def solicitar_pres():
  print("===SOLICITAR PRESTAMO===")
  num_control = input("Número de control del alumno: ")
  if num_control in alumnos:
      programa = input("¿Qué programa necesita? ")
      for i in range(len(computadoras)):
          if programa.lower() in computadoras[i][2].lower():
              if computadoras[i][3] == "Disponible":
                        computadoras[i][3] = "Prestada"
                        print("Se asignó la computadora", computadoras[i][0],
                              "que esta en el aula", computadoras[i][1])
                        break
              else:
                  print("Si se encontro el sofware pero esta", computadoras[i][3])
  else:
    print("El alumno no existe en la lista")


#4
def devolver_computadora():
  print("\n ===DEVOLVER COMPUTADORA===")
  id_pc=input("Dame el ID de la computadora: ")
  existe = False
  for i in range(len(computadoras)):
      if computadoras[i][0] ==id_pc:
        existe = True
        if computadoras[i][3]=="Prestada" or computadoras[i][3]=="Reservada":
          computadoras[i][3] = "Disponible"
          print("La computadora " ,id_pc,"ya esta disponible")
          break
        else:
          print("La computadora no esta prestada/reservada")
  if existe==False:
    print("No existe la computadora", id_pc)

#6
def agregar_computadora():
  print("\n--- AGREGAR COMPUTADORA ---")
  id_computadora = input("ID de la computadora: ")
  aula = input("Aula: ")
  programas = input("Software instalado: ")
  nueva_compu=[id_computadora ,aula , programas ,"Disponible"]
  computadoras.append(nueva_compu)
  print("\nComputadora agregada correctamente.")


# MENÚ PRINCIPAL
# Usamos un ciclo While que se repetiria indefinidamente hasta que el ussuario
# teclee la opcio  7 (salir) o cualquier numero superior a 7
opcion = 0
while opcion <= 6:
    print("\n=================================")
    print("        SISTEMA DE PRÉSTAMOS   ")
    print("===============================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar préstamo")
    print("4. Devolver computadora")
    print("5. Crear reservación")
    print("6. Agregar computadora")
    print("7. Salir")
    print("===============================")
    try:
        opcion = int(input("Seleccione una opción: "))
    except ValueError:
        print("Entrada inválida. Por favor, ingrese un número.")
        continue

    #1. Consultar computadoras
    # si el usuario me selecciono la opcion 1 se hace un ciclo (for) que reconoce
    # toda la lista de computadoras e imprime los valores de las demás listas
    if opcion == 1:
        listar_computadoras() # Mandamos a llamar la función (def listar_computadoras) que desarrollamos antes.

    # 2. Buscar por software
    elif opcion == 2:
        buscar_sw() #Al mandar a llamar la funcion ponemos solo el titulo de esta función

    #3. Solicitar préstamo
    elif opcion == 3:
        solicitar_pres()

    # 4.
    elif opcion == 4:
      devolver_computadora()


    # 6. Agregar computadora
    elif opcion == 6:
     agregar_computadora()
JKLJ
```
